## Cài đặt thư viện nltk

```python
%pip install nltk requests
import nltk
from nltk.metrics import distance
```

```python
import string
import unicodedata
import requests
```

## Lấy văn bản từ Wikipedia

```python
url = "https://vi.wikipedia.org/w/api.php"
params = {"action": "query", "prop": "extracts", "explaintext": 1,
          "titles": "Việt Nam", "format": "json", "redirects": 1}
headers = {"User-Agent": "UEH-NLP-exercise"}

data = requests.get(url, params = params, headers = headers).json()
text = list(data["query"]["pages"].values())[0]["extract"]
len(text)
print(text)
```

```python
text[:500]
```

## a. Tách từ không quan tâm ngữ nghĩa

Tách theo khoảng trắng (mỗi token là một âm tiết), không xét từ ghép.

```python
text = unicodedata.normalize("NFC", text).lower()

tokens = text.split()
len(tokens)
print(tokens)
```

```python
tokens[:30]
```

## b. Loại bỏ các từ không đạt chuẩn (số, ký hiệu đặc biệt, emoji)

```python
punct = string.punctuation + "“”‘’…–—«»"

tokens = [w.strip(punct) for w in tokens]
tokens = [w for w in tokens if w.isalpha()]
len(tokens)
```

```python
tokens[:30]
```

## c. Bộ từ điển và tần số xuất hiện của từng từ

```python
unigram = nltk.FreqDist(tokens)

vocab = sorted(unigram)
print(vocab)
len(vocab)
```

```python
unigram.most_common(20)
```

## d. Bigrams: top 10

```python
bigram = nltk.FreqDist(nltk.bigrams(tokens))
bigram.most_common(10)
```

## e. Trigrams: top 10

```python
trigram = nltk.FreqDist(nltk.trigrams(tokens))
trigram.most_common(10)
```

## f. Kiểm tra các cặp từ có khả năng của phiên bản sai "diệt nam"

- Với mỗi từ của cụm sai, tìm các từ trong bộ từ điển có edit distance ≤ 1
- Ghép thành các cặp và chỉ giữ cặp có xuất hiện trong bigrams của corpus
- Nếu không có cặp nào tồn tại thì tăng edit distance lên 2

```python
wrong = ["diệt", "nam"]

for d in [1, 2]:
    candidates = [[w for w in vocab if distance.edit_distance(x, w, substitution_cost = 1) <= d] for x in wrong]
    pairs = [(a, b) for a in candidates[0] for b in candidates[1] if (a, b) in bigram]
    print("edit distance <=", d, ":", len(pairs), "cặp tồn tại trong corpus")
    if pairs:
        break
```

```python
for a, b in sorted(pairs, key = lambda p: -bigram[p])[:10]:
    med = distance.edit_distance(" ".join(wrong), a + " " + b, substitution_cost = 1)
    print(f"{a} {b:<10} tần số = {bigram[(a, b)]:<5} MED với 'diệt nam' = {med}")
```
