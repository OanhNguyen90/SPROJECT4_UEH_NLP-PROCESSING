## 2. Mã hóa và áp dụng thuật toán Minimum Edit Distance cho cấp bậc câu

Dữ liệu đầu vào

```python
source1 = "Tôi học tiếng anh"
target2 = "Tôi yêu tiếng việt"
```

Bước 1: Tách từ của s, t

```python
source_tokens = source1.lower().split()
target_tokens = target2.lower().split()

print(source_tokens)
print(target_tokens)
```

Bước 2: Tạo bộ từ điển D = S ∪ T

```python
D = sorted(set(source_tokens) | set(target_tokens))
print(D)
```

Bước 3: Mã hóa bộ từ điển D

```python
# Tạo ra một cấu trúc dữ liệu Dictionary giúp tra cứu ID từ một từ: {'học'; 0} trong đó 'học' là Key; 0 là Index
word_to_id = {word: idx for idx, word in enumerate(D)}

# Tạo ra một Dictionary ngược lại để tra cứu ngược từ ID ra lại từ gốc: {0; 'học'} trong đó 0 là Index còn 'học' là Key
id_to_word = {idx: word for word, idx in word_to_id.items()} 
print(word_to_id)
```

Bước 4: Dùng bộ từ điển mã hóa để mã hóa s và t

```python
# Mã hóa câu nguồn (source câu) từ dạng chữ sang dạng số.
source_encoded = [word_to_id[w] for w in source_tokens]

# Mã hóa câu đích (target câu - thường là câu kết quả hoặc câu dịch) từ dạng chữ sang dạng số.
target_encoded = [word_to_id[w] for w in target_tokens]

print(source_encoded)
print(target_encoded)
```

Bước 5: Tính MED(s, t) với chi phí thay thế bằng 1

```python
distance.edit_distance(source_encoded, target_encoded, substitution_cost = 1)
```

MED(s, t) với chi phí thay thế bằng 2 (giống phần 1)

```python
distance.edit_distance(source_encoded, target_encoded, substitution_cost = 2)
```

Căn chỉnh (alignment) giữa hai câu

```python
distance.edit_distance_align(source_encoded, target_encoded, substitution_cost = 2)
```
