Quy ước mặc định (trừ khi đề nói khác): chèn = 1, xoá = 1, thay thế = 2, thay ký tự giống nhau = 0.

````python
# Hàm hỗ trợ kiểm tra - chạy ô này trước, KHÔNG cần sửa
def check(name, got, expected):
    if got == expected:
        print(f"✅ {name}")
        return True
    print(f"❌ {name}: nhận được {got!r}, kỳ vọng {expected!r}")
    return False

def run_tests(tests):
    results = []
    for name, f in tests:
        try:
            results.append(check(name, *f()))
        except NotImplementedError:
            print(f"⏳ {name}: chưa làm (còn NotImplementedError)"); results.append(False)
        except Exception as e:
            print(f"💥 {name}: lỗi {type(e).__name__}: {e}"); results.append(False)
    print(f"\nKết quả: {sum(results)}/{len(results)} test đạt", "🎉" if all(results) else "")
````

````python
def build_with_pointers(source, target):
    n, m = len(source), len(target)
    D = [[0] * (m + 1) for _ in range(n + 1)]
    B = [[set() for _ in range(m + 1)] for _ in range(n + 1)]
    for i in range(1, n + 1): D[i][0] = i; B[i][0].add("up")
    for j in range(1, m + 1): D[0][j] = j; B[0][j].add("left")
    for i in range(1, n + 1):
        for j in range(1, m + 1):
            s = 0 if source[i-1] == target[j-1] else 2
            up, left, diag = D[i-1][j] + 1, D[i][j-1] + 1, D[i-1][j-1] + s
            best = min(up, left, diag)
            D[i][j] = best
            if up == best: B[i][j].add("up")
            if left == best: B[i][j].add("left")
            if diag == best: B[i][j].add("diag")
    return D, B
````

**Đáp án Bài 9 (⭐⭐⭐)**

````python
def align(source, target):
    D, B = build_with_pointers(source, target)
    i, j, ops = len(source), len(target), []
    while i > 0 or j > 0:
        if "diag" in B[i][j]:
            a, b = source[i-1], target[j-1]
            ops.append(("keep" if a == b else "sub", a, b)); i -= 1; j -= 1
        elif "up" in B[i][j]:
            ops.append(("del", source[i-1], None)); i -= 1
        else:
            ops.append(("ins", None, target[j-1])); j -= 1
    return ops[::-1]
````

````python
rows = ["qwertyuiop", "asdfghjkl", "zxcvbnm"]
pos = {ch: (r, c) for r, row in enumerate(rows) for c, ch in enumerate(row)}
def adjacent(a, b):
    if a not in pos or b not in pos: return False
    (r1, c1), (r2, c2) = pos[a], pos[b]
    return abs(r1 - r2) <= 1 and abs(c1 - c2) <= 1

def _kdist(s, t):
    n, m = len(s), len(t)
    D = [[0] * (m + 1) for _ in range(n + 1)]
    for i in range(1, n + 1): D[i][0] = i
    for j in range(1, m + 1): D[0][j] = j
    for i in range(1, n + 1):
        for j in range(1, m + 1):
            a, b = s[i-1], t[j-1]
            sc = 0 if a == b else (1 if adjacent(a, b) else 2)
            D[i][j] = min(D[i-1][j] + 1, D[i][j-1] + 1, D[i-1][j-1] + sc)
    return D[n][m]

def suggest(typo, words, k=3):
    return sorted((_kdist(typo, w), w) for w in words)[:k]
````

````python
def med_low_memory(s, t):
    if len(s) < len(t): s, t = t, s          # t là chuỗi ngắn
    prev = list(range(len(t) + 1))
    for i in range(1, len(s) + 1):
        cur = [i] + [0] * len(t)
        for j in range(1, len(t) + 1):
            sc = 0 if s[i-1] == t[j-1] else 2
            cur[j] = min(prev[j] + 1, cur[j-1] + 1, prev[j-1] + sc)
        prev = cur
    return prev[-1]
````

````python
INF = float("inf")
def within_k(s, t, k):
    n, m = len(s), len(t)
    if abs(n - m) > k: return None
    prev = {j: j for j in range(0, min(m, k) + 1)}            # hàng i = 0
    for i in range(1, n + 1):
        cur = {}
        lo, hi = max(0, i - k), min(m, i + k)
        for j in range(lo, hi + 1):
            if j == 0:
                cur[j] = i; continue
            sc = 0 if s[i-1] == t[j-1] else 2
            cur[j] = min(prev.get(j, INF) + 1, cur.get(j-1, INF) + 1, prev.get(j-1, INF) + sc)
        prev = cur
    d = prev.get(m, INF)
    return d if d <= k else None
````

**Đáp án Bài 17 (⭐⭐⭐⭐)**

````python
import math
P_MATCH, P_INS, P_DEL, P_SUB = 0.9, 0.02, 0.02, 0.01

def best_alignment_prob(s, t):
    n, m = len(s), len(t)
    ci, cd, cs, cm = -math.log(P_INS), -math.log(P_DEL), -math.log(P_SUB), -math.log(P_MATCH)
    D = [[0.0] * (m + 1) for _ in range(n + 1)]
    for i in range(1, n + 1): D[i][0] = D[i-1][0] + cd
    for j in range(1, m + 1): D[0][j] = D[0][j-1] + ci
    for i in range(1, n + 1):
        for j in range(1, m + 1):
            sub = cm if s[i-1] == t[j-1] else cs
            D[i][j] = min(D[i-1][j] + cd, D[i][j-1] + ci, D[i-1][j-1] + sub)
    return math.exp(-D[n][m])

def most_likely_source(typo, words):
    return max(sorted(words), key=lambda w: best_alignment_prob(w, typo))
````
