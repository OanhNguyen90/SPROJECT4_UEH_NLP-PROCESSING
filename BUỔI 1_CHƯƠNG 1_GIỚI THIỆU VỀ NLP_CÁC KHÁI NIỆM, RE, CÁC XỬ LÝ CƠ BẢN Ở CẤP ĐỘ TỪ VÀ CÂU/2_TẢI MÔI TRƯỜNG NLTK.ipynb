```python

import sys
import subprocess
import importlib
import os

print("=" * 50)
print("1. KIỂM TRA MÔI TRƯỜNG PYTHON")
print("=" * 50)
print("Python:", sys.version)
print("Executable:", sys.executable)

print("\n" + "=" * 50)
print("2. GỠ VÀ CÀI ĐẶT LẠI NLTK")
print("=" * 50)

subprocess.run([
    sys.executable, "-m", "pip", "uninstall", "-y", "nltk"
], check=False)

result = subprocess.run([
    sys.executable, "-m", "pip", "install",
    "--no-cache-dir", "--force-reinstall", "nltk"
], check=False)

if result.returncode != 0:
    raise RuntimeError(
        "Cài đặt NLTK thất bại. Hãy kiểm tra kết nối mạng và pip."
    )

print("\n" + "=" * 50)
print("3. NẠP LẠI MODULE")
print("=" * 50)

importlib.invalidate_caches()
sys.modules.pop("nltk", None)

import nltk

print("Module file:", getattr(nltk, "__file__", None))
print("Version:", getattr(nltk, "__version__", "Unknown"))
print("Has download:", hasattr(nltk, "download"))

if not hasattr(nltk, "download"):
    raise ImportError(
        "NLTK vẫn chưa được nạp đúng. "
        "Hãy kiểm tra file nltk.py hoặc thư mục nltk "
        "trong thư mục dự án, sau đó restart kernel."
    )

print("\n" + "=" * 50)
print("4. TẢI DỮ LIỆU NLTK")
print("=" * 50)

nltk.download("punkt", quiet=False)
nltk.download("punkt_tab", quiet=False)

print("\nHoàn tất! NLTK đã sẵn sàng.")
```
