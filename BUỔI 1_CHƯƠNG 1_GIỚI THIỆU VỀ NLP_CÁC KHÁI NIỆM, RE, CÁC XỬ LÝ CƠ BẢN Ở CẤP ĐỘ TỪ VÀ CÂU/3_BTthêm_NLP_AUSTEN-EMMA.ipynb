# Buổi 1 – Chương 1: Xử lý văn bản cơ bản với NLTK
Dữ liệu: corpus `austen-emma.txt` của NLTK

## Câu 1. Sử dụng corpus của nltk với file austen-emma

```python
import re
%pip install nltk
import nltk

nltk.download('gutenberg')
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.corpus import gutenberg

text = gutenberg.raw('austen-emma.txt')
print(len(text))
print(text[:300])
```

## Câu 2. Tách các từ có chứa chữ 'g'

```python
words_g = re.findall(r'\b\w*g\w*\b', text)
print(len(words_g))
print(words_g[:20])
```

## Câu 3. Tách các từ có đuôi '-able' (không có ký tự nào theo sau)

```python
words_able = re.findall(r'\b\w+able\b', text)
print(len(words_able))
print(set(words_able))
```

## Câu 4. Tách các từ có chứa '-ation', phía trước và phía sau vẫn có ký tự

```python
words_ation = re.findall(r'\b\w+ation\w+\b', text)
print(len(words_ation))
print(set(words_ation))
```

## Câu 5. Tách từ (tokenize)

```python
from nltk.tokenize import word_tokenize

tokens = word_tokenize(text)
print(len(tokens))
print(tokens[:20])
```

## Câu 6. Tách câu

```python
from nltk.tokenize import sent_tokenize

sentences = sent_tokenize(text)
print(len(sentences))
print(sentences[:3])
```

## Câu 7. Tìm gốc từ bằng Porter Stemmer
Lưu ý: stemming không phải lúc nào cũng đúng hoặc ra từ có nghĩa; kết quả là dạng gốc (stem) chứ không phải nguyên thể (lemma).

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

# chỉ lấy các từ là chữ cái, bỏ dấu câu
words = [t.lower() for t in tokens if t.isalpha()]

# loại trùng để bảng kết quả gọn hơn
unique_words = sorted(set(words))

stems = [(w, stemmer.stem(w)) for w in unique_words]
print(len(stems))
print(stems[:30])
```

Ví dụ các từ stem ra kết quả không có nghĩa (vd: `emma` → `emma`, `beautiful` → `beauti`).

```python
for w in ['beautiful', 'happiness', 'considerable', 'generally', 'was']:
    print(w, '->', stemmer.stem(w))
```
